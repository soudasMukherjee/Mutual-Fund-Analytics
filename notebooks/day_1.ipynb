```python

nav_history = pd.read_csv(data_dir / "02_nav_history.csv")

key_schemes = [119551, 120503, 118632, 119092, 120841]
key_navs = (
    nav_history.loc[nav_history["amfi_code"].isin(key_schemes)]
    .assign(date=lambda frame: pd.to_datetime(frame["date"]))
    .sort_values(["amfi_code", "date"])
    .groupby("amfi_code", as_index=False)
    .tail(1)
    .merge(fund_master[["amfi_code", "scheme_name"]], on="amfi_code", how="left")
    [["amfi_code", "scheme_name", "date", "nav"]]
    .sort_values("amfi_code")
    .reset_index(drop=True)
)
print(key_navs)
```

```python
print('Unique fund houses:')
print(sorted(fund_master['fund_house'].dropna().unique()))

print('\nUnique categories:')
print(sorted(fund_master['category'].dropna().unique()))

print('\nUnique sub-categories:')
print(sorted(fund_master['sub_category'].dropna().unique()))

print('\nUnique risk grades:')
print(sorted(fund_master['risk_category'].dropna().unique()))

fund_master_sorted = fund_master.sort_values('amfi_code').reset_index(drop=True)
print('\nAMFI scheme code mapping:')
print(fund_master_sorted[['amfi_code', 'scheme_name', 'fund_house', 'plan']].to_string(index=False))

codes = fund_master_sorted['amfi_code'].tolist()
contiguous_groups = []
group_start = codes[0]
previous_code = codes[0]

for code in codes[1:]:
    if code == previous_code + 1:
        previous_code = code
    else:
        contiguous_groups.append((group_start, previous_code, previous_code - group_start + 1))
        group_start = previous_code = code

contiguous_groups.append((group_start, previous_code, previous_code - group_start + 1))

print('\nContiguous AMFI code blocks (start, end, size):')
print(contiguous_groups)
```

```python
fund_codes = set(fund_master['amfi_code'])
nav_codes = set(nav_history['amfi_code'])

missing_in_nav = sorted(fund_codes - nav_codes)
extra_in_nav = sorted(nav_codes - fund_codes)

print(f'Fund master codes: {len(fund_codes)}')
print(f'NAV history codes: {len(nav_codes)}')
print(f'Codes missing in NAV history: {missing_in_nav}')
print(f'Codes present only in NAV history: {extra_in_nav}')

if not missing_in_nav:
    print('Validation passed: every fund_master AMFI code exists in nav_history.')
else:
    print('Validation failed: some fund_master codes are missing from nav_history.')
```
